```python
import pandas as pd 

df = pd.read_csv('comparison/test_data/sz000001_RJJ.csv')
```

```python
order_id = df['ORDER_ID']
#查看是否有重复
print(order_id.duplicated().sum())
#查看重复的order_id
print(order_id[order_id.duplicated()])
print(len(order_id))
print(len(order_id[order_id.duplicated()]))


#将所有重复的order_id的行保存为新的csv文件
duplicate_orders = df[df['ORDER_ID'].duplicated(keep=False)]
duplicate_orders.to_csv('comparison/test_data/duplicate_orders.csv', index=False)


```

```python
# 检查order_id相同的行，除了TIMESTAMP和SIMUTIME，其他列是否完全相同
cols_to_check = [col for col in df.columns if col not in ['TIMESTAMP', 'SIMUTIME']]
duplicates = df[df.duplicated(subset=cols_to_check, keep=False)]
# 按order_id排序，方便查看
duplicates = duplicates.sort_values(by='ORDER_ID')
duplicates.to_csv('comparison/test_data/fully_duplicate_rows.csv', index=False)
```

```python
dup_ids = df[df.duplicated(subset=['ORDER_ID'], keep=False)]\
            .sort_values(['ORDER_ID', 'TIMESTAMP'])

changes = dup_ids.groupby('ORDER_ID')[cols_to_check].nunique()

# 查看哪些列发生过变化
changed_cols = changes[(changes > 1).any(axis=1)]
```

```python
# 将原始数据中的order_id去重（保留第一次出现的），然后保存为新的csv文件
df = pd.read_csv('comparison/test_data/sz000001_RJJ.csv')
df_unique = df.drop_duplicates(subset=['ORDER_ID'], keep='first')
df_unique.to_csv('comparison/test_data/sz000001_unique.csv', index=False)
```
